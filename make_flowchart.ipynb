```python
from pyflowchart import *

start = StartNode("SAT")

read_dxf = OperationNode("Read in DXF and visualise")
edit_structure = OperationNode("Edit joint types / structure")
read_materials = OperationNode("Read in materials sheet")
assign_materials = OperationNode("Assign member materials")

check_det = ConditionNode("Structure determinate?")

apply_loads = OperationNode("Define and apply loads")
check_eq = OperationNode("Check equilibrium")
solve = OperationNode("Solve axial forces and bending moments")
calc_stresses = OperationNode("Calculate stresses")
optimise = OperationNode("Optimise the structure")

end = EndNode("Write DXF to edit in AutoCAD")

# Main workflow
start.connect(read_dxf)
read_dxf.connect(edit_structure)
edit_structure.connect(read_materials)
read_materials.connect(assign_materials)
assign_materials.connect(check_det)

# Determinacy loop
check_det.connect_yes(apply_loads)
check_det.connect_no(edit_structure)

# Analysis workflow
apply_loads.connect(check_eq)
check_eq.connect(solve)
solve.connect(calc_stresses)
calc_stresses.connect(optimise)
optimise.connect(end)

print(Flowchart(start).flowchart())
```
